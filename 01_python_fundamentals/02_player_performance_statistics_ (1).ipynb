# Player Performance Statistics with Python

## Objective

This notebook explores basic football performance statistics using Python.

## Topics covered

- Working with lists and dictionaries.
- Creating reusable functions.
- Calculating goals and assists per 90 minutes.
- Using loops to process multiple players.
- Storing results in structured dictionaries.
- Sorting players to create performance rankings.

## Dataset

A fictional dataset containing four football players, their minutes played, goals and assists.

## Tools

Python – Google Colab.

```python
players = [
    {
        "name": "Player A",
        "position": "Forward",
        "minutes": 1800,
        "goals": 12,
        "assists": 5
    },
    {
        "name": "Player B",
        "position": "Forward",
        "minutes": 2100,
        "goals": 10,
        "assists": 8
    },
    {
        "name": "Player C",
        "position": "Midfielder",
        "minutes": 1900,
        "goals": 4,
        "assists": 9
    },
    {
        "name": "Player D",
        "position": "Forward",
        "minutes": 900,
        "goals": 7,
        "assists": 2
    }
]
```

## 2. Exploring the dataset

Before calculating player statistics, let's explore the dataset and understand how to access the information stored in each dictionary.

```python
for player in players:
  print(player)
```

```python
for player in players:
  if player["position"]=='Forward':
    print(player["name"])
```

## 3. Calculating player statistics

```python
player_stats = []
def calculate_assists_per_90(assists, minutes):
  formula = (assists/minutes)*90
  return formula
def calculate_goals_per_90(goals, minutes):
  formula = (goals/minutes)*90
  return formula
for player in players:

    assists_per_90 = calculate_assists_per_90(
        player["assists"], player["minutes"]
    )

    goals_per_90 = calculate_goals_per_90(
        player["goals"], player["minutes"]
    )

    player_stats.append({
        "name": player["name"],
       "goals_per_90": goals_per_90,
        "assists_per_90": assists_per_90
    })
print(player_stats)
```

## 4. Creating player rankings

```python
ranking_goals=sorted(
    player_stats,
    key = lambda player: player["goals_per_90"],
    reverse = True
)
```

```python
print(ranking_goals)
```

```python
ranking_assists = sorted(
    player_stats,
    key = lambda player: player["assists_per_90"],
    reverse = True
)
```

```python
print(ranking_assists)
```

```python
for player in ranking_goals:
  print(f'{player["name"]}: {player["goals_per_90"]} goals per 90 minutes')
```

```python
print("\nAssists per 90 ranking:")

for player in ranking_assists:
    print(f'{player["name"]}: {player["assists_per_90"]:.2f} assists per 90 minutes')
```
